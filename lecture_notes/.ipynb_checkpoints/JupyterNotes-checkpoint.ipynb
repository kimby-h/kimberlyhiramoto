```python
import numpy as np
list1 = [1, 2, 3]
list2 = [4, 5, 6]
print(list1 + list2)
list1, list2 = np.array([list1]), np.array([list2])
print(list1 + list2)
```

```python
list = [20, 40, 60, 80, 100]
def divide(list):
    for num in list:
        print(num/5)
divide(list)
```

```python
array = np.array([20, 40, 60, 80, 100])
print(array/5)
```
